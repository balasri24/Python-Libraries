```python
penguins=sns.load_dataset('penguins')
penguins.head()
```

```python
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="sex")  # insted of sex we can use island also # hue tells that which penguin belongs to which island
```

```python
# the ways of styling themes are as follows:
# white
# dark
# whitegrid
# darkgrid
# ticks


# set_context() allows us to override default parameters.This affects things like the size of the labels,lines , and other elements of the plot , but not the overall style ->set_style().the context are:
# poster- very large elements -suitable for large displays or posters
# Paper -Ideal for printing in acedamic papers.(smallest elements)
# Notebook - Good for working inside jupyter notebooks (default setting).(medium size)
# Talk - good for presentation slides where readability mattters.
```

```python
sns.set_style("darkgrid") # insted of ticks we can use white,dark,whitegrid,darkgrid,ticks
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
sns.despine( left=True) # the name at the top is removed
```

```python
sns.set_style("darkgrid")
sns.set_context("poster")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
```

```python
sns.set_style("darkgrid")
sns.set_context("talk") # talk decreses the size as compered to poster
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
```

```python
sns.set_style("darkgrid")
sns.set_context("paper")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
```

```python
# palette        description                                     max Distinct Colors
 # set1           bright and high- contrast colors                 9
 # set2           softter, more pastel-like colors                  8
 # set3           a mix of soft and bold ,varied hues               12

 #"deep"          defalut palette,rich and balenced
 #"muted"         softer colors ,good for many categories
 #"bright"        high-contrast,bold colors
 #"dark"          darker tones,easy to distinguish
 #"colorblind"    optimized for colorblind accessibility

# palette type        purpose                               examples                           best used for
#(qualitative)     (categorical in nature, the color      'set1',set2,set3,pastel1,Dark2,      distinguishing categories or groups
                   #assigned to each group need to be      colorblind,accent
                   #distinct)
#sequential        data with ordered progression           Blues ,Greens , Purples,OrRd,       showing magnitude or intensity,like population or revenue
                  #(low to high)                          YIGnBu,BuPu, Greys
#diverging         # data with a meningful midpoint(eg.,0)  coolwarm,RdBu,BrBG,PiYG,Spectral,   showing posive vs negative change(eg.,profi/loss),temperature change
                   #or mixed value like +ve and -ve          terrain_r
                    #(low and high values)
```
