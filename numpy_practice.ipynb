```python
import numpy as np
arr=np.array([1,2,3,4,5])
```

```python
# 1st question
print(arr)
arr[2]=100
print("updated array",arr)
```

```python
# 2nd question
a=np.eye(4)
print(a)
b=np.eye(4,k=2)
print(b)
```

```python
# 3rd question
a=np.array([5,10,25,20,25,30])
print(a[::2]) # index based even
print(a[1::2]) # position based even
```

```python
# 4th question
# create random 3X3 array
a=np.random.randint(10,51,size=(3,3))
print(a)
# extract last colum
print(a[:,2])
# or
print(a[:,-1])
```

```python
# 5th question
a=np.array([[12,24],[36,48],[60,72]])
# to print totla number of elements
print("total number of elements",a.size)
# shape
print("shape ",a.shape)
# dimensions
print("dimensions",a.ndim)
```

```python
# 6th question
# create 1D array of 12 elements  using np,arange()
a=np.arange(12) # or np.arange(0,12)
print(a)
# reshape to 3x4
print(a.reshape(3,4))
# (or) in one line
b=np.arange(12).reshape(3,4)
print(b)
```

```python
# 7th question
d=np.array([[5,15],[25,35],[45,55]])
print(d)
# transpose of this matrix using .T
a=d.T
print(a)
# modify any element in transepose matrix
a[0,1]=500
print(a)
print(d) # so transpose matrix  shares memory with original matrix

```

```python
# 8th question
# create an array of 6 elements
z=np.array([1,2,3,4,5,6])
print(z)
# resize it into 2x3 matrix
z.resize(2,3)
print(z)
# resize to a large shape 3x3 matrix
z.resize(3,3)
print(z)
```

```python
# 9th question
d=np.array([15,22,8,19,33,4,27])
# greater then 15 using boolean indexing
print(d[d>15])
```

```python
# 10th question
# 3x3 array with all elements set to 7 using np.full()
a=np.full((3,3),7)
print(a)
# transpose
b=a.T
print(b)
```
