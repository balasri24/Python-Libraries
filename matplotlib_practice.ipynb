```python
import matplotlib.pyplot as plt
#import pandas as pd
#import numpy as np
RK=[8,7,13,24,32,50]
tier1=["IIT Dhanbad(Jarkhand)","III Roorkee(uttarkhand)","IIT Madras(chenni)","III kharagpur(wesbengal)","IIT Guwanhat(assam)","IIT Hydrabad(Telangana)"]
tier2=["iitdm(kurnool)","iiit kalyani(westbengal)","iiit luknow","iiit sricity(a.p)","iiit hydrabad(telangana)","iiit(ongole,a.p)"]
plt.plot(tier1,RK,marker="v",linestyle="-.",label="IIT") # :, -. ,_ ,__ dable dash
plt.plot(tier2,RK,color="red",label="IIIT") # use legend for lable
plt.xticks(rotation=45)
plt.legend()
plt.title("NIRF of IIT and IIIT Colleges")
plt.xlabel("college rankings")
plt.ylabel("ranking positions")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
a=[10,20,30,40,50,60,70]
b=["sun","mon","tue","wed","thu","fri","sat"]
plt.plot(a,b)
plt.title("weekly temperature")
plt.xlabel("temperature")
plt.ylabel("days")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
a=[10,20,30,40,50,60,70]
b=["sun","mon","tue","wed","thu","fri","sat"]
plt.bar(a,b)
plt.title("weekly temperature")
plt.xlabel("temperature")
plt.ylabel("days")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
a=["stickers","clips","bangles","dresses","ear rings","ice cream","watches"]
b=[10,20,100,400,50,200,250]
plt.scatter(a,b,color=["red","black","yellow","blue","gold","pink","white"]) # bar , barh,pie,scatter
plt.title("purchases of items")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
items=["crab","prawns","fish","octupus","star fish","chicken","mutton"]
price=[350,200,300,800,700,320,1000]
plt.pie(price,labels=items,autopct="%1.1f%%")
plt.title("weekly purchase items",color="red")
plt.show()
```
