```python
# spliting
import torch
a=torch.tensor([1,10,11,12,15,16])
print(torch.split(a,5))
print(torch.split(a,4))
b=torch.arange(20)
print(torch.split(b,[3,7,5,4,1]))
print(torch.chunk(b,3))
c=torch.zeros(4,5,6)
print(c.size(1))
c=torch.ones(4,5,6)
print(c.size(1))
```

```python
# operations
import torch
a=torch.tensor([[5,6,7],[8,9,10]])
print(a.sum())
print(a.sum(dim=1))
print(a.sum(dim=0))
print(a.sum(axis=0))
print(a.sum(axis=1))
print(a.mean(dim=0,dtype=float))
print(a.mean(dim=1,dtype=float))
print(a.max())
print(a.max(dim=1))
print(a.max(dim=0))
```

```python
b=torch.tensor([24,36,48,100,56,57])
print(b.size())
print(b.argmax())
print(b.argmin())
```
