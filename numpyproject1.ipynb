```python
import numpy as np
marks=np.array([[14 ,24 , 11 , 23 , 17 ],[ 17, 16 , 15 , 16 ,19 ],[ 20 , 19 , 18 , 20 ,11 ],[ 14 , 13 , 12 , 14 , 19 ]
                ,[ 11 , 10 , 9 , 17 , 20 ],[14 ,24 ,11 , 34 , 56 ],[ 14, 13 , 12 , 35 , 47],[ 17, 16 , 15 , 19 , 76 ],
                 [ 88 , 87 , 84 , 85 , 86 ],[ 14 , 13 , 12 ,  17 , 74 ]])
subjects=["Maths","Python","c","Telugu","English"]
print("\nStudent Marks:")
print("          Maths Python C  Telugu English ")
for i in range(10):
  print("Student",i+1," ",marks[i][0]," ",marks[i][1]," ",marks[i][2]," ",marks[i][3]," ",marks[i][4] )
#students=np.array([" student 1"," student 2"," student 3"," student 4"," student 5"," student 6"," student 7"," student 8"," student 9"," student 10",])
```

```python
# total marks of each student
print("total marks of each student:",np.sum(marks,axis=1))
```

```python
# average marks of each student
average=np.mean(marks,axis=1)
print("average marks of each student:",average)
```

```python
# hight marks
print("highest marks:",np.max(marks,axis=1))
```

```python
# lowest marks
print("highest marks:",np.min(marks))
```

```python
# student with highest average
averages=np.mean(marks,axis=1)
highest_idx=np.argmax(averages)
print("Highest Average Student:",highest_idx+1)
```

```python
# subject wise average
average=np.mean(marks,axis=0)
print("subject wise average :",average)
```

```python
# student who scored above 80 in all subjects
for i in range(10):
  if np.all(marks[i]>80):
    print(f"Student {i+1}")
```

```python
# subject wise average
average=np.mean(marks,axis=0)
print("subject wise average:",average)
```

```python
# student who failed in any subject
for i in range(10):
  if np.any(marks[i]<35):
    print(f" Student{i+1}")
```

```python
# Rank of student based on total marks
totals=list(np.sum(marks,axis=1))
for rank in range(1,11):
  highest_mark=max(totals)
  student_index=totals.index(highest_mark)
  print(f"Rank {rank}:Student {student_index +1} (Total:{highest_mark})")
  totals[student_index]=-1
```

```python
# class average and standard deviation
class_average=np.mean(marks)
class_std_dev=np.std(marks)
print("class average marks:",class_average)
print("class standard deviation:",class_std_dev)
```

```python
# normalize marks
min_val=np.min(marks)
max_val=np.max(marks)
normalized_marks=(marks-min_val)/(max_val-min_val)
#print(normalized_marks)
print(np.round(normalized_marks,2))
```

```python
student_averages=np.mean(marks,axis=1)
for i in range(10):
  avg=student_averages[i]
  if avg >=90:
    category="excellent"
  elif(75<=avg<=89):
    category="good"
  elif(50<=avg<75):
    category="average"
  else:
    category="poor"
  print(f"Student {i+1} -> Average: {avg} -> {category}")
```
