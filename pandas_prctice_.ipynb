```python
df["promoted salary"]=df["Salary"]*10
df
```

```python
# data cleaning
df.isnull().sum()
```

```python
df.dropna() # it removes the null value row
```

```python
df.dropna(how="any") # defalul is any
```

```python
df.dropna(how="all") # if all the values in my row are null then we remove thet row
```

```python
df.fillna(0)  # we get the output as 0.0[a value isthe aa value NaN  place lo replace avuthundhi] in the place of NaN
```

```python
df.fillna(2)
```

```python
df["Age"].fillna(df["Age"].mean()) # the null value is replaced with the average of age values (25+30+35+29+25)/5=28.8
```

```python
df["Salary"].fillna(df["Salary"].median())
```

```python
df["Age"].fillna(method="ffill")  #forward fill# null value is replaced by the before value [means null value gets before value]
```

```python
df["Age"].fillna(method="bfill") # backward fill# null value is replaced by the after value [means null value gets after value]
```

```python
df["Name"].replace("srija","sri")
```

```python
# dealing with duplicate values
df_dup=df[df.duplicated()]
df_dup
```

```python
df_dup=df[df.duplicated(keep="last")] # last duplicate value is not treated as duplicate value
df_dup
```

```python
df_dup=df[df.duplicated(keep="first")]  # it is defult # first value not treated as duplicate value
df_dup
```

```python
df=df.drop_duplicates() # remove duplicate value row
df
```

```python
df["promoted salary"]=df["promoted salary"].apply(lambda x:x/10 if x>650000 else x)
df
```

```python
# apply and lambda
def multiply_age(x):
  return x*2
df["Age"]=df["Age"].apply(multiply_age)
df
```

```python
df["Age"]=df["Age"].apply(lambda x:x/2)
df
```

```python
# joins and merges
department_info={
    "Dept":["HR","IT","Finance"],
    "location":["New york","san Francisco","chicago"],
    "Manager":["laura","steve","nina"]
}
df2=pd.DataFrame(department_info)
df2
```

```python
# concat
pd.concat([df,df2])
```

```python
pd.concat([df,df2],axis=1)
```

```python
pd.merge(df,df2,on="Dept") # if the name is same in both dataframe then we simple use on or otherwise we use left_on="1st dataframe lo name",right_on=" 2nd dataframe lo name"
```
